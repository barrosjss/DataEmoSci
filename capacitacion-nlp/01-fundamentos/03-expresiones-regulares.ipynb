# Expresiones regulares

Las expresiones regulares (a veces llamadas regex para abreviar) permiten al usuario buscar cadenas utilizando casi cualquier tipo de regla que se le ocurra. Por ejemplo, encontrar todas las letras mayúsculas en una cadena o encontrar un número de teléfono en un documento.

Las expresiones regulares son notorias por su sintaxis aparentemente extraña. Esta extraña sintaxis es un subproducto de su flexibilidad. Las expresiones regulares tienen que ser capaces de filtrar cualquier patrón de cadena que puedas imaginar, razón por la cual tienen un formato de patrón de cadena complejo.

Las expresiones regulares se manejan usando la biblioteca **re** incorporada de Python. Consulte [la documentación](https://docs.python.org/3/library/re.html) para obtener más información.

¡Comencemos explicando cómo buscar patrones básicos en una cadena!

## Búsqueda de patrones básicos

Imaginemos que tenemos la siguiente cadena:

```python
text = "The agent's phone number is 408-555-1234. Call soon!"
```

Comenzaremos tratando de averiguar si la cadena "phone" está dentro de la cadena de texto. Ahora podríamos hacer esto rápidamente con:

```python
'phone' in text
```

Pero mostremos el formato de las expresiones regulares, porque más adelante estaremos buscando patrones que no tendrán una solución tan simple.

```python
import re
```

```python
pattern = 'phone'
```

```python
re.search(pattern,text)
```

```python
pattern = "NOT IN TEXT"
```

```python
re.search(pattern,text)
```

Ahora hemos visto que re.search() tomará el patrón, escaneará el texto y luego devolverá un objeto Match. Si no se encuentra ningún patrón, se devuelve None (en Jupyter Notebook esto solo significa que no se genera nada debajo de la celda).

Echemos un vistazo más de cerca a este objeto Match.

```python
pattern = 'phone'
```

```python
match = re.search(pattern,text)
```

```python
match
```

Observe que en span, también hay una información de índice de inicio y final.

```python
match.span()
```

```python
match.start()
```

```python
match.end()
```

Pero, ¿y si el patrón ocurre más de una vez?

```python
text = "my phone is a new phone"
```

```python
match = re.search("phone",text)
```

```python
match.span()
```

Observe que solo coincide con la primera instancia. Si quisiéramos una lista de todas las coincidencias, podemos usar el método .findall():

```python
matches = re.findall("phone",text)
```

```python
matches
```

```python
len(matches)
```

Para obtener todos los objetos coincidentes, use el iterador:

```python
for match in re.finditer("phone",text):
    print(match.span())
```

Si se quiere el texto coincidente, puede usar el método .group().

```python
match.group()
```

# Patrones

Hasta ahora hemos aprendido cómo buscar una cadena básica. ¿Qué pasa con ejemplos más complejos? ¿Como tratar de encontrar un número de teléfono en una gran cadena de texto? ¿O una dirección de correo electrónico?

Podríamos usar el método de búsqueda si conocemos el teléfono o el correo electrónico exactos, pero ¿y si no lo sabemos? Es posible que conozcamos el formato general, y podemos usarlo junto con expresiones regulares para buscar en el documento cadenas que coincidan con un patrón particular.

Aquí es donde la sintaxis puede parecer extraña al principio, pero tómese su tiempo con esto; a menudo es solo cuestión de buscar el código del patrón.

## Identificadores de caracteres en patrones

Los caracteres como un dígito o una sola cadena tienen diferentes códigos que los representan. Puede usarlos para construir una cadena de patrón. Observe cómo estos hacen un uso intensivo de la barra invertida \ . Debido a esto, cuando definimos una cadena de patrón para una expresión regular, usamos el formato:

     r'mipatrón'
    
colocar la r delante de la cadena le permite a python entender que la \ en la cadena del patrón no está destinada a ser barras de escape.

A continuación puede encontrar una tabla con todos los posibles identificadores:

<table ><tr><th>Caracter</th><th>Descripcíon</th><th>Ejemplo Patrón</th><th >Ejemplo concidencia</th></tr>

<tr ><td><span >\d</span></td><td>A digit</td><td>file_\d\d</td><td>file_25</td></tr>

<tr ><td><span >\w</span></td><td>Alphanumeric</td><td>\w-\w\w\w</td><td>A-b_1</td></tr>



<tr ><td><span >\s</span></td><td>White space</td><td>a\sb\sc</td><td>a b c</td></tr>



<tr ><td><span >\D</span></td><td>A non digit</td><td>\D\D\D</td><td>ABC</td></tr>

<tr ><td><span >\W</span></td><td>Non-alphanumeric</td><td>\W\W\W\W\W</td><td>*-+=)</td></tr>

<tr ><td><span >\S</span></td><td>Non-whitespace</td><td>\S\S\S\S</td><td>Yoyo</td></tr></table>

Por ejemplo:

```python
text = "My telephone number is 408-555-1234"
```

```python
phone = re.search(r'\d\d\d-\d\d\d-\d\d\d\d',text)
```

```python
phone.group()
```

Note la repetición de \d. Eso es un poco molesto, especialmente si estamos buscando cadenas de números muy largas. Exploremos los posibles cuantificadores.

## Cuantificadores

Ahora que conocemos las designaciones de caracteres especiales, podemos usarlas junto con cuantificadores para definir cuántos esperamos.

Reescribamos nuestro patrón usando estos cuantificadores:

```python
re.search(r'\d{3}-\d{3}-\d{4}',text)
```

## Grupos

¿Qué pasaría si quisiéramos hacer dos tareas, encontrar números de teléfono, pero también poder extraer rápidamente su código de área (los primeros tres dígitos). Podemos usar grupos para cualquier tarea general que implique agrupar expresiones regulares (para luego desglosarlas).

Usando el ejemplo del número de teléfono, podemos separar grupos de expresiones regulares usando paréntesis:

```python
phone_pattern = re.compile(r'(\d{3})-(\d{3})-(\d{4})')
```

```python
results = re.search(phone_pattern,text)
```

```python
# Todo el resultado
results.group()
```

```python
# También puede llamar por posición de grupo.
# recuerda que los grupos estaban separados por paréntesis ()
# Algo a tener en cuenta es que el pedido de grupo comienza en 1. Pasar 0 devuelve todo
results.group(1)
```

```python
results.group(2)
```

```python
results.group(3)
```

```python
# Solo teníamos tres grupos de paréntesis.
results.group(4)
```

## Sintaxis Regex adicional

### O operador |

Use el operador de canalización para tener una declaración **o**. Por ejemplo

```python
re.search(r"man|woman","This man was here.")
```

```python
re.search(r"man|woman","This woman was here.")
```

### El carácter comodín (wildcard)

Utilice un "comodín" como ubicación que coincidirá con cualquier carácter colocado allí. Puede usar un punto simple **.** para esto. Por ejemplo:

```python
re.findall(r".at","The cat in the hat sat here.")
```

```python
re.findall(r".at","The bat went splat")
```

Observe cómo solo hicimos coincidir las primeras 3 letras, eso es porque necesitamos un **.** para cada letra comodín o use los cuantificadores descritos anteriormente para establecer sus propias reglas.

```python
re.findall(r"...at","The bat went splat")
```

Sin embargo, esto todavía conduce al problema de agarrar más de antemano. Realmente solo queremos palabras que terminen con "at".

```python
# Uno o más espacios sin blancos que terminan con 'at'
re.findall(r'\S+at',"The bat went splat")
```

### Comienza con y termina con

Podemos usar **^** para señalar que comienza con, y **$** para señalar que termina con:

```python
# Termina con un número
re.findall(r'\d$','This ends with a number 2')
```

```python
# Comienza con un número
re.findall(r'^\d','1 is the loneliest number.')
```

¡Tenga en cuenta que esto es para toda la cadena, no para palabras individuales!

### Exclusión

Para excluir caracteres, podemos usar el símbolo **^** junto con un conjunto de corchetes **[]**. Se excluye todo lo que esté dentro de los corchetes. Por ejemplo:

```python
phrase = "there are 3 numbers 34 inside 5 this sentence."
```

```python
re.findall(r'[^\d]',phrase)
```

Para volver a juntar las palabras, use un signo +

```python
re.findall(r'[^\d]+',phrase)
```

Podemos usar esto para eliminar la puntuación de una oración.

```python
test_phrase = 'This is a string! But it has punctuation. How can we remove it?'
```

```python
re.findall('[^!.? ]+',test_phrase)
```

```python
clean = ' '.join(re.findall('[^!.? ]+',test_phrase))
```

```python
clean
```

## Corchetes para agrupar

Como mostramos arriba, podemos usar corchetes para agrupar opciones, por ejemplo, si quisiéramos encontrar palabras con guión:

```python
text = 'Only find the hypen-words in this sentence. But you do not know how long-ish they are'
```

```python
re.findall(r'[\w]+-[\w]+',text)
```

## Paréntesis para opciones múltiples

Si tenemos varias opciones para hacer coincidir, podemos usar paréntesis para enumerar estas opciones. Por ejemplo:

```python
# Encuentra palabras que comiencen con cat y terminen con una de estas opciones: 'fish', 'nap' o 'claw'
text = 'Hello, would you like some catfish?'
texttwo = "Hello, would you like to take a catnap?"
textthree = "Hello, have you seen this caterpillar?"
```

```python
re.search(r'cat(fish|nap|claw)',text)
```

```python
re.search(r'cat(fish|nap|claw)',texttwo)
```

```python
# None returned
re.search(r'cat(fish|nap|claw)',textthree)
```
