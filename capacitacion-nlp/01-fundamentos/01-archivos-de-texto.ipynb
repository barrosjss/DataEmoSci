## Python abriendo un archivo

Para evitar este error, asegúrese de que su archivo .txt esté guardado en la misma ubicación que su computadora portátil. Para verificar la ubicación de su computadora portátil, use **pwd**:

**Alternativamente, para capturar archivos desde cualquier ubicación en su computadora, simplemente ingrese la ruta completa del archivo. **

Para Windows, debe usar doble \ para que Python no trate el segundo \ como un carácter de escape, una ruta de archivo tiene la forma:

Para MacOS y Linux, usa barras en la dirección opuesta:

```python
# Abra el archivo text.txt que creamos anteriormente
my_file = open('test.txt')
```

```python
my_file
```

`my_file` ahora es un objeto de archivo abierto guardado en la memoria. Realizaremos algunos ejercicios de lectura y escritura, y luego tendremos que cerrar el archivo para liberar memoria.


### .read() and .seek()

```python
# Ahora podemos leer el archivo.
my_file.read()
```

```python
# Pero, ¿qué pasa si tratamos de leerlo de nuevo?
my_file.read()
```

Esto sucede porque puedes imaginar que el "cursor" de lectura está al final del archivo después de haberlo leído. Así que no queda nada por leer. Podemos restablecer el "cursor" así:

```python
# Buscar el inicio del archivo (índice 0)
my_file.seek(0)
```

```python
# Ahora lee de nuevo
my_file.read()
```

### .readlines()
Puede leer un archivo línea por línea usando el método readlines. Tenga cuidado con los archivos grandes, ya que todo se mantendrá en la memoria. Aprenderemos a iterar sobre archivos grandes más adelante en el curso.

```python
# Readlines devuelve una lista de las líneas en el archivo
my_file.seek(0)
my_file.readlines()
```

Cuando haya terminado de usar un archivo, siempre es una buena práctica cerrarlo.

```python
my_file.close()
```

## Escribir en un archivo

Por defecto, la función `open()` solo nos permitirá leer el archivo. Necesitamos pasar el argumento `'w'` para escribir sobre el archivo. Por ejemplo:

```python
# Agregue un segundo argumento a la función, 'w', que significa escribir.
# Pasar 'w+' nos permite leer y escribir en el archivo

my_file = open('test.txt','w+')
```

<div class="alert alert-danger" style="margin: 20px">**¡Ten cuidado!**<br>
Abrir un archivo con 'w' o 'w+' *trunca el original*, lo que significa que todo lo que estaba en el archivo original **se elimina**.</div>

```python
# Escribir en el archivo
my_file.write('This is a new first line')
```

```python
# Leer el archivo
my_file.seek(0)
my_file.read()
```

```python
my_file.close()  # siempre haga esto cuando este listo con un archivo
```

## Agregar a un archivo
Al pasar el argumento `'a'` se abre el archivo y se coloca el puntero al final, por lo que se agrega todo lo escrito. Al igual que `'w+'`, `'a+'` nos permite leer y escribir en un archivo. Si el archivo no existe, se creará uno.

```python
my_file = open('test.txt','a+')
my_file.write('\nThis line is being appended to test.txt')
my_file.write('\nAnd another line here.')
```

```python
my_file.seek(0)
print(my_file.read())
```

```python
my_file.close()
```

### Agregar con `%%writefile`
Los usuarios de notebooks de Jupyter pueden hacer lo mismo usando la magia de celda IPython (comándos mágicos):

```python
%%writefile -a test.txt

This is more text being appended to test.txt
And another line here.
```

Agregue un espacio en blanco si desea que la primera línea comience en su propia línea, ya que Jupyter no reconocerá secuencias de escape como `\n`

## Aliases y Context Managers
Puede asignar nombres de variables temporales como alias y administrar la apertura y el cierre de archivos automáticamente mediante un administrador de contexto:

```python
with open('test.txt','r') as txt:
    first_line = txt.readlines()[0]
    
print(first_line)
```

Tenga en cuenta que el administrador de contexto `with ... as ...:` cerró automáticamente `test.txt` después de asignar la primera línea de texto a first_line:

```python
txt.read()
```

## Iterando a través de un archivo

```python
with open('test.txt','r') as txt:
    for line in txt:
        print(line, end='')  # el end='' argumento remueve extra linebreaks
```
