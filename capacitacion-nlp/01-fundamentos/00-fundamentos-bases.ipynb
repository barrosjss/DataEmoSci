## Literales de cadena con formato (f-strings)

Introducido en Python 3.6, <strong>f-strings</strong> ofrece varios beneficios sobre el antiguo método de cadena `.format()`. <br>Por un lado, puede traer variables externas inmediatamente a la cadena en lugar de pasarlas como argumentos de palabras clave:

```python
name = 'Fred'

# Usando el viejo método .format():
print('His name is {var}.'.format(var=name))

# Usando f-strings:
print(f'His name is {name}.')
```

Pase `!r` para obtener la <strong>representación de cadena</strong>:

```python
print(f'His name is {name!r}')
```

Tenga cuidado de no dejar que las comillas en los campos de reemplazo entren en conflicto con las comillas utilizadas en la cadena exterior:

```python
d = {'a':123,'b':456}

print(f'Address: {d['a']} Main Street')
```

En su lugar, utilice diferentes estilos de comillas:

```python
d = {'a':123,'b':456}

print(f"Address: {d['a']} Main Street")
```

### Anchos mínimos, alineación y relleno
Puede pasar argumentos dentro de un conjunto anidado de llaves para establecer un ancho mínimo para el campo, la alineación e incluso los caracteres de relleno.

```python
library = [('Author', 'Topic', 'Pages'), ('Twain', 'Rafting', 601), ('Feynman', 'Physics', 95), ('Hamilton', 'Mythology', 144)]

for book in library:
    print(f'{book[0]:{10}} {book[1]:{8}} {book[2]:{7}}')
```

Aquí las tres primeras líneas se alinean, excepto que "Páginas" sigue una alineación a la izquierda predeterminada, mientras que los números se alinean a la derecha. Además, el número de página de la cuarta línea se desplaza hacia la derecha, ya que "Mitología" supera el ancho de campo mínimo de "8". Al establecer anchos de campo mínimos, asegúrese de tener en cuenta el elemento más largo.

Para establecer la alineación, use el carácter `<` para alinear a la izquierda, `^` para el centro, `>` para la derecha.<br>
Para configurar el relleno, preceda el carácter de alineación con el carácter de relleno (`-` y `.` son opciones comunes).

Hagamos algunos ajustes:

```python
for book in library:
    print(f'{book[0]:{10}} {book[1]:{10}} {book[2]:.>{7}}') # here .> was added
```

### Formato de fecha

```python
from datetime import datetime

today = datetime(year=2018, month=1, day=27)

print(f'{today:%B %d, %Y}')
```
