# Semántica y Vectores de Palabras
A veces llamado "minería de opinión", [Wikipedia](https://en.wikipedia.org/wiki/Sentiment_analysis) define ***análisis de sentimiento*** como
<div class="alert alert-info" style="margin: 20px">"el uso del procesamiento del lenguaje natural... para identificar, extraer, cuantificar y estudiar sistemáticamente estados afectivos e información subjetiva.<br>
En términos generales, el análisis de sentimientos tiene como objetivo determinar la actitud de un orador, escritor u otro tema con respecto a algún tema o la polaridad contextual general o la reacción emocional a un documento, interacción o evento".</div>

Hasta ahora, hemos utilizado la ocurrencia de palabras específicas y patrones de palabras para realizar clasificaciones de prueba. En esta sección, llevaremos el aprendizaje automático aún más lejos e intentaremos extraer los significados previstos de frases complejas. Algunos ejemplos simples incluyen:
* Python es relativamente fácil de aprender.
* Esa fue la peor película que he visto.

Sin embargo, las cosas se ponen más difíciles con frases como:
* No me disgustan los huevos verdes y el jamón. (requiere manejo de negación)

La forma en que esto se hace es a través de complejos algoritmos de aprendizaje automático como [word2vec](https://en.wikipedia.org/wiki/Word2vec). La idea es crear matrices numéricas o *incrustaciones de palabras* para cada palabra en un corpus grande. A cada palabra se le asigna su propio vector de tal manera que a las palabras que frecuentemente aparecen juntas en el mismo contexto se les asignan vectores que están muy juntos. El resultado es un modelo que puede no saber que un "león" es un animal, pero sí sabe que "león" está más cerca en contexto de "gato" que de "diente de león".

Es importante tener en cuenta que *construir* modelos útiles lleva mucho tiempo (horas o días para entrenar un corpus grande) y que para nuestros propósitos es mejor importar un modelo existente en lugar de tomarse el tiempo para entrenar el nuestro.

# Vectores de palabras
Los vectores de palabras, también llamados *word embeddings*, son descripciones matemáticas de palabras individuales, de modo que las palabras que aparecen juntas con frecuencia en el idioma tendrán valores similares. De esta manera podemos derivar matemáticamente *contexto*. Como se mencionó anteriormente, el vector de palabra para "león" tendrá un valor más cercano a "gato" que a "diente de león".

## Valores vectoriales
Entonces, ¿cómo se ve un vector de palabras? Dado que spaCy emplea 300 dimensiones, los vectores de palabras se almacenan como matrices de 300 elementos.

Tenga en cuenta que veríamos el mismo conjunto de valores con **en_core_web_md** y **en_core_web_lg**, ya que ambos fueron entrenados usando la familia de algoritmos [word2vec](https://en.wikipedia.org/wiki/Word2vec) .

```python
# Importar spaCy y cargar la biblioteca de idiomas
import spacy
nlp = spacy.load('es_core_news_md')  # ¡asegúrate de usar un modelo más grande!
```

```python
nlp(u'león').vector
```

Lo que es interesante es que los objetos Doc y Span serían vectores en si, derivados de los promedios de los vectores de tokens individuales. <br>Esto hace posible comparar similitudes entre documentos completos.

```python
doc = nlp(u'El astuto zorro marrón saltó por encima de los peresozos perros.')

doc.vector
```

## Identificar vectores similares
La mejor forma de exponer las relaciones vectoriales es a través del método `.similarity()` de los tokens Doc.

```python
# Cree un objeto Doc de tres tokens:
tokens = nlp(u'león gato mascota')

# Iterar a través de combinaciones de tokens:
for token1 in tokens:
    for token2 in tokens:
        print(token1.text, token2.text, token1.similarity(token2))
```

<font color=green>Tenga en cuenta que el orden no importa. `token1.similarity(token2)` tiene el mismo valor que `token2.similarity(token1)`.</font>
#### Para ver esto como una tabla:

```python
# Por brevedad, asigne un nombre a cada token
a,b,c = tokens

# Mostrar como una tabla Markdown (¡esto solo funciona en Jupyter!)
from IPython.display import Markdown, display
display(Markdown(f'<table><tr><th></th><th>{a.text}</th><th>{b.text}</th><th>{c.text}</th></tr>\
<tr><td>**{a.text}**</td><td>{a.similarity(a):{.4}}</td><td>{b.similarity(a):{.4}}</td><td>{c.similarity(a):{.4}}</td></tr>\
<tr><td>**{b.text}**</td><td>{a.similarity(b):{.4}}</td><td>{b.similarity(b):{.4}}</td><td>{c.similarity(b):{.4}}</td></tr>\
<tr><td>**{c.text}**</td><td>{a.similarity(c):{.4}}</td><td>{b.similarity(c):{.4}}</td><td>{c.similarity(c):{.4}}</td></tr>'))
```

Como era de esperar, vemos la mayor similitud entre "gato" y "mascota", la más débil entre "león" y "mascota", y cierta similitud entre "león" y "gato". Una palabra tendrá una similitud perfecta (1.0) consigo misma.

Si tienes curiosidad, la similitud entre "león" y "pantaleon" es muy pequeña:

```python
nlp(u'león').similarity(nlp(u'pantaleon'))
```

### Los opuestos no son necesariamente diferentes
Las palabras que tienen significados opuestos, pero que a menudo aparecen en el mismo *contexto*, pueden tener vectores similares.

```python
# Cree un objeto Doc de tres tokens:
tokens = nlp(u'comer rezar amar')

# Iterate through token combinations:
for token1 in tokens:
    for token2 in tokens:
        print(token1.text, token2.text, token1.similarity(token2))
```

## Norma de vectores
A veces es útil agregar 300 dimensiones en una [norma euclidiana (L2)] (https://en.wikipedia.org/wiki/Norm_%28mathematics%29#Euclidian_norm), calculada como la raíz cuadrada de la suma de cuadrados -vectores. Esto es accesible como el atributo de token `.vector_norm`. Otros atributos útiles incluyen `.has_vector` y `.is_oov` o *out of vocabulary*.

Por ejemplo, nuestra biblioteca de vectores puede no tener la palabra "[abudinar](https://www.infobae.com/america/colombia/2021/09/07/la-rae-aclaro-su-posicion-sobre-las-palabras-abudinar-y-abudinear/)". Para probar esto:

```python
tokens = nlp(u'comer rezar abudinear')

for token in tokens:
    print(token.text, token.has_vector, token.vector_norm, token.is_oov)
```

De hecho, vemos que "abudinear" no tiene un vector, por lo que el valor vector_norm es cero y se identifica como *fuera de vocabulario*.

## Aritmética vectorial
Lo creas o no, podemos calcular nuevos vectores sumando y restando vectores relacionados. Un ejemplo famoso sugiere
<pre>"rey" - "hombre" + "mujer" = "reina"</pre>
¡Vamos a probarlo con alguna profesión!

```python
from scipy import spatial

cosine_similarity = lambda x, y: 1 - spatial.distance.cosine(x, y)

king = nlp.vocab['doctor'].vector
man = nlp.vocab['hombre'].vector
woman = nlp.vocab['mujer'].vector

# Now we find the closest vector in the vocabulary to the result of "man" - "woman" + "queen"
new_vector = king - man + woman
computed_similarities = []

for word in nlp.vocab:
    # Ignore words without vectors and mixed-case words:
    if word.has_vector:
        if word.is_lower:
            if word.is_alpha:
                similarity = cosine_similarity(new_vector, word.vector)
                computed_similarities.append((word, similarity))

computed_similarities = sorted(computed_similarities, key=lambda item: -item[1])

print([w[0].text for w in computed_similarities[:10]])
```

Entonces, en nuestro caso, "rey", no fue mu concluyente, pero si lo hacemos con alguna profesión los resulatdos son bastante satisfactorios
