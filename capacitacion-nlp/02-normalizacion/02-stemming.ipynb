# Stemming (Derivación)
A menudo, al buscar texto para una determinada palabra clave, ayuda si la búsqueda arroja variaciones de la palabra. Por ejemplo, la búsqueda de "barco" también puede dar como resultado "barcos" y "barca". Aquí, "barco" sería la **raíz** de [barco, barca, barcos].

Stemming es un método un tanto tosco para catalogar palabras relacionadas; esencialmente corta las letras desde el final hasta que se alcanza la raíz. Esto funciona bastante bien en la mayoría de los casos, pero desafortunadamente dependiendo del idioma tiene muchas excepciones donde se requiere un proceso más sofisticado. De hecho, spaCy no incluye un lematizador, sino que opta por confiar completamente en la lematización. Para aquellos interesados, hay algunos antecedentes sobre esta decisión [aquí] (https://github.com/explosion/spaCy/issues/327). Discutimos las virtudes de la *lematización* en la siguiente sección.

En su lugar, usaremos otra herramienta popular de NLP llamada **nltk**, que significa *Natural Language Toolkit*. Para obtener más información sobre nltk, visite https://www.nltk.org/

## Porter Stemmer

Una de las herramientas de stemming más comunes y efectivas es el [*Algoritmo de Porter*](https://tartarus.org/martin/PorterStemmer/) desarrollado por Martin Porter en [1980](https://tartarus.org/martin /PorterStemmer/def.txt). El algoritmo emplea cinco fases de reducción de palabras, cada una con su propio conjunto de reglas de mapeo. En la primera fase, se definen reglas simples de mapeo de sufijos, tales como:

![stemming1.png](stemming1.png)

De un conjunto dado de reglas derivadas, solo se aplica una regla, basada en el sufijo S1 más largo. Así, `caress` se reduce a `caress` pero no a `cares`.

Las fases más sofisticadas consideran la longitud/complejidad de la palabra antes de aplicar una regla. Por ejemplo:

![stemming1.png](stemming2.png)

Aquí `m>0` describe la "medida" de la raíz, de modo que la regla se aplica a todas las raíces excepto a las más básicas.

```python
# Importe el kit de herramientas y la biblioteca completa de Porter Stemmer
import nltk

from nltk.stem.porter import *
```

```python
p_stemmer = PorterStemmer()
```

```python
words = ['run','runner','running','ran','runs','easily','fairly']
```

```python
for word in words:
    print(word+' --> '+p_stemmer.stem(word))
```

<font color=green>Observe cómo el lematizador reconoce "runner" como un sustantivo, no como una forma verbal o participio. Además, los adverbios "easily" y "fairly" se derivan de la raíz inusual "easili" y "fairli"</font>

## Snowball Stemmer
Este es un nombre algo inapropiado, ya que Snowball es el nombre de un lenguaje derivado desarrollado por Martin Porter. El algoritmo que se utiliza aquí se denomina con mayor precisión "English Stemmer" o "Porter2 Stemmer". Ofrece una ligera mejora sobre el stemmer de Porter original, tanto en lógica como en velocidad. Dado que **nltk** usa el nombre SnowballStemmer, lo usaremos aquí.

```python
from nltk.stem.snowball import SnowballStemmer

# El Snowball Stemmer requiere que pase un parámetro de idioma
s_stemmer = SnowballStemmer(language='english')
```

```python
words = ['run','runner','running','ran','runs','easily','fairly']
# words = ['generous','generation','generously','generate']
```

```python
for word in words:
    print(word+' --> '+s_stemmer.stem(word))
```

<font color=green>En este caso, el lematizador se desempeñó igual que el Porter Stemmer, con la excepción de que manejó la raíz de "fairly" más apropiadamente con "fair"</font>

___
Stemming tiene sus inconvenientes. Si se le da el token `saw`, stemming siempre puede devolver `saw`, mientras que la lematización probablemente devolverá `see` o `saw` dependiendo de si el token se usa como verbo o como sustantivo. Como ejemplo, considere lo siguiente:

```python
phrase = 'I am meeting him tomorrow at the meeting'
for word in phrase.split():
    print(word+' --> '+p_stemmer.stem(word))
```

Aquí la palabra "reunión" aparece dos veces: una como verbo y otra como sustantivo, y, sin embargo, el lematizador trata a ambos por igual.
