# Taller de conceptos básicos de NLP - Soluciones

Para esta evaluación, utilizaremos el cuento [_An Occurrence at Owl Creek Bridge_](https://en.wikipedia.org/wiki/An_Occurrence_at_Owl_Creek_Bridge) de Ambrose Bierce (1890). <br>La historia es de dominio público; el archivo de texto se obtuvo del [Proyecto Gutenberg](https://www.gutenberg.org/ebooks/375.txt.utf-8).

```python
# EJECUTE ESTA CELDA para realizar importaciones estándar:
import spacy
nlp = spacy.load('en_core_web_sm')
```

**1. Cree un objeto Doc desde el archivo `owlcreek.txt`**<br>
> SUGERENCIA: Usa `with open('../datos/owlcreek.txt') como f:`

```python
# Introduce tu código aquí:

with open('../datos/owlcreek.txt') as f:
    doc = nlp(f.read())
```

```python
# Ejecute esta celda para verificar que funcionó:

doc[:36]
```

**2. ¿Cuántos tokens contiene el archivo?**

```python
len(doc)
```

**3. ¿Cuántas oraciones contiene el archivo?**<br>PISTA: ¡Primero querrá crear una lista!

```python
sents = [sent for sent in doc.sents]
len(sents)
```

**4. Imprima la segunda oración en el documento**<br> SUGERENCIA: La indexación comienza en cero y el título cuenta como la primera oración.

```python
print(sents[1].text)
```

** 5. Para cada token en la oración anterior, imprima su `texto`, etiqueta `POS`, etiqueta `dep` y `lemma`<br>
RETO: Hacer que los valores se alineen en columnas en la salida impresa.**

```python
# SOLUCIÓN NORMAL:
for token in sents[1]:
    print(token.text, token.pos_, token.dep_, token.lemma_)
```

```python
# SOLUCIÓN MENOS NORMAL:
for token in sents[1]:
    print(f'{token.text:{15}} {token.pos_:{5}} {token.dep_:{10}} {token.lemma_:{15}}')
```

**6. Escriba un matcher llamado 'Swimming' que encuentre ambas apariciones de la frase "swimming vigorously" en el texto**<br>
SUGERENCIA: Debes incluir un patrón `'IS_SPACE': True` entre las dos palabras.

```python
# Importar la biblioteca Matcher:
from spacy.matcher import Matcher
matcher = Matcher(nlp.vocab)
```

```python
# Cree un patrón y agréguelo al matcher:

pattern = [{'LOWER': 'swimming'}, {'IS_SPACE': True, 'OP':'*'}, {'LOWER': 'vigorously'}]

matcher.add('Swimming', None, pattern)
```

```python
# Cree una lista de coincidencias llamada "found_matches" e imprima la lista:

found_matches = matcher(doc)
print(found_matches)
```

**7. Imprime el texto que rodea cada coincidencia encontrada**

```python
print(doc[1265:1290])
```

```python
print(doc[3600:3615])
```

**EXTRA PUNTOS:<br>Imprime la *frase* que contiene cada coincidencia encontrada**

```python
for sent in sents:
    if found_matches[0][1] < sent.end:
        print(sent)
        break
```

```python
for sent in sents:
    if found_matches[1][1] < sent.end:
        print(sent)
        break
```
