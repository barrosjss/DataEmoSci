# Lematización
A diferencia de stemming, la lematización va más allá de la reducción de palabras y considera el vocabulario completo de un idioma para aplicar un *análisis morfológico* a las palabras. El lema de 'was' es 'be' y el lema de 'mice' es 'mouse'. Además, el lema de 'reunión' podría ser 'reunirse' o 'reunión' dependiendo de su uso en una oración.

```python
# Realizar importaciones estándar:
import spacy
nlp = spacy.load('en_core_web_sm')
```

```python
doc1 = nlp(u"I am a runner running in a race because I love to run since I ran today")

for token in doc1:
    print(token.text, token.pos_, token.lemma, token.lemma_, sep='\t')
```

<font color=green>En la oración anterior, `running`, `run` y `ran` apuntan al mismo lema `run` (...11841) para evitar la duplicación.</font>

### Función para mostrar lemas
Dado que la pantalla de arriba está escalonada y es difícil de leer, escribamos una función que muestre la información que queremos de manera más clara.

```python
def show_lemmas(text):
    for token in text:
        print(f'{token.text:{12}} {token.pos_:{6}} {token.lemma:<{22}} {token.lemma_}')
```

Aquí estamos usando una **f-string** para formatear el texto impreso estableciendo anchos de campo mínimos y agregando una alineación a la izquierda al valor hash del lema.

```python
doc2 = nlp(u"I saw eighteen mice today!")

show_lemmas(doc2)
```

<font color=green>Observe que el lema de 'saw' es 'see', 'mice' es la forma plural de 'mouse' y, sin embargo, 'eighteen' es su propio número, *no* una forma expandida de 'eight' `.</font>

```python
doc3 = nlp(u"I am meeting him tomorrow at the meeting.")

show_lemmas(doc3)
```

<font color=green>Aquí el lema de `meeting` está determinado por su etiqueta de parte del discurso (POS).</font>

```python
doc4 = nlp(u"That's an enormous automobile")

show_lemmas(doc4)
```

<font color=green>Tenga en cuenta que la lematización *no* reduce las palabras a su sinónimo más básico, es decir, "enormous" no se convierte en "big" y "automobile" no se convierte en "car".</font>

Debemos señalar que aunque la lematización observa el texto circundante para determinar la parte del discurso de una palabra dada, no clasifica las frases. En una próxima lección investigaremos *vectores de palabras y similitudes*.
