# Para las palabras
Palabras como "a" y "the" aparecen con tanta frecuencia que no requieren un etiquetado tan exhaustivo como los sustantivos, los verbos y los modificadores. Las llamamos *palabras vacías* y se pueden filtrar del texto que se va a procesar. spaCy tiene una lista integrada de unas 305 palabras vacías en inglés.

```python
# Realizar importaciones estándar:
import spacy
nlp = spacy.load('en_core_web_sm')
```

```python
# Imprima el conjunto de palabras vacías predeterminadas de spaCy (recuerde que los conjuntos no están ordenados):
print(nlp.Defaults.stop_words)
```

```python
len(nlp.Defaults.stop_words)
```

## Para ver si una palabra es una stop word

```python
nlp.vocab['myself'].is_stop
```

```python
nlp.vocab['mystery'].is_stop
```

## Para agregar una stop word
Puede haber ocasiones en las que desee agregar una palabra vacía al conjunto predeterminado. Tal vez decidas que `'btw'` (abreviatura común para "by the way") debe considerarse una palabra vacía.

```python
# Add the word to the set of stop words. Use lowercase!
nlp.Defaults.stop_words.add('btw')

# Set the stop_word tag on the lexeme
nlp.vocab['btw'].is_stop = True
```

```python
len(nlp.Defaults.stop_words)
```

```python
nlp.vocab['btw'].is_stop
```

<font color=green>Al agregar palabras vacías, use siempre minúsculas. Los lexemas se convierten a minúsculas antes de agregarse al **vocabulario**.</font>

## Para eliminar una stop word
Alternativamente, puede decidir que `'beyond'` no debe considerarse una palabra vacía.

```python
# Eliminar la palabra del conjunto de stop words
nlp.Defaults.stop_words.remove('beyond')

# Remover stop_words tag del lexema
nlp.vocab['beyond'].is_stop = False
```

```python
len(nlp.Defaults.stop_words)
```

```python
nlp.vocab['beyond'].is_stop
```

¡Excelente! Ahora debería poder acceder al conjunto predeterminado de stop words de spaCy y agregar o eliminar stop words según sea necesario.
